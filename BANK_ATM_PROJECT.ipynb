# BANK ATM SIMULATION SYSTEM

```python
print("~"*140)
print(" "*60,"AXIS BANK")  
print("~"*140)
print()
print()
print("1.Cash Withdraw"," "*15,"2.Cash Deposit"," "*15,"3.Balance Check"," "*15,"4.Change Pin"," "*15,"5.Exit")
print()
print()
Pin=int(input("Enter 4 digit permanent pin for your account: "))

Balance=int(input("Enter your bank balance: "))

def Withdraw():
    global Balance
    P=int(input("Enter 4 digit pin : "))
    if P==Pin:
        Amt=int(input("Enter amount for withdraw: "))
        if Amt<=Balance:
            Balance=Balance-Amt
        else:
            print("Insufficient funds.")
    else:
        print("Enter correct pin.")
        
def Deposit():
    global Balance
    P=int(input("Enter 4 digit pin : "))
    if P==Pin:
        Dept=int(input("Enter amount for deposit: "))
        Balance=Balance+Dept
    else:
        print("Enter correct pin.")

def Check():
    global Balance
    P=int(input("Enter 4 digit pin : "))
    if P==Pin:
        print("Your bank balance is: ",Balance)
    else:
        print("Enter correct pin.")

def New_Pin():
    global Pin
    Old=int(input("Enter old pin: "))
    if Old==Pin:
        New=int(input("Enter new pin: "))
        ReNew=int(input("Re-enter new pin:  "))
        if ReNew==New:
            Pin=New
            print("Pin changed succesfully.")
        else:
            print("Enter correct pin.")
    else:
        print("Enter correct pin.")

while True:
    Choice=int(input("Enter your choice(1-5) : "))
    if Choice==1:
        Withdraw()
    elif Choice==2:
        Deposit()
    elif Choice==3:
        Check()
    elif Choice==4:
        New_Pin()
    elif Choice==5:
        ans=input("Are you sure you want to exit ?")
        if ans in "YesYESyes":
            print("Have a nice day.")
            break
    else:
        break
```
